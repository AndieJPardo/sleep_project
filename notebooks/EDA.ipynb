```python
import pandas as pd
import plotly_express as px
```

```python
df = pd.read_csv("../sleepdata.csv")
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

```python
fig = px.histogram(
    df,
    x="Sleep_Hours",
    nbins=30,
    title="Distribución de Horas de Sueño"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Daily_Phone_Hours",
    y="Sleep_Hours",
    color="Stress_Level",
    title="Daily Phone Hours vs Sleep Hours"
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Sleep_Hours",
    y="Stress_Level",
    trendline="ols",
    title="Sleep Hours vs Stress Level"
)

fig.show()
```

```python
fig = px.box(
    df,
    x="Gender",
    y="Stress_Level",
    title="Stress Level by Gender"
)

fig.show()
```

```python
occupation_stress = df.groupby("Occupation")["Stress_Level"].mean().reset_index()

fig = px.bar(
    occupation_stress,
    x="Occupation",
    y="Stress_Level",
    title="Average Stress Level by Occupation"
)

fig.show()
```
